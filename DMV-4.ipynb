```python
import pandas as pd
import matplotlib.pyplot as plt

try:
    df = pd.read_csv("RealEstate_Prices.csv")
    print("Successfully loaded 'RealEstate_Prices.csv' from local storage.")
except FileNotFoundError:
    print("Local file 'RealEstate_Prices.csv' not found. Creating a sample dataset for demonstration...")
    data = {
        "Sale Price": [350000, 420000, 310000, 550000, 275000, 480000, 390000, 610000, 450000, 330000],
        "Sale Date": ["2025-01-15", "2025-01-18", "2025-02-01", "2025-02-10", "2025-02-14", "2025-03-02", "2025-03-05", "2025-03-12", "2025-03-20", "2025-03-25"],
        "Neighborhood": ["Downtown", "Suburbs", "Downtown", "Northhills", "Suburbs", "Northhills", "Downtown", "Northhills", "Suburbs", "Downtown"],
        "Square_Feet": [1800, 2200, 1500, 2800, 1400, 2500, 1900, 3100, 2300, 1600]
    }
    df = pd.DataFrame(data)
df.columns = (df.columns.str.strip().str.lower()
              .str.replace(r"[^a-z0-9]+", "_", regex=True).str.strip("_"))

print("\nInitial DataFrame Head:")
print(df.head())
print("\nMissing values:\n", df.isna().sum())

if "saleprice" in df.columns:
    df["saleprice"] = pd.to_numeric(df["saleprice"], errors="coerce")
    df = df.dropna(subset=["saleprice"])
    df = df[df["saleprice"] > 0]

if "saledate" in df.columns:
    df["saledate"] = pd.to_datetime(df["saledate"], errors="coerce")

if "neighborhood" in df.columns and "saleprice" in df.columns:
    summary = df.groupby("neighborhood")["saleprice"].agg(
        ["count", "mean", "median", "min", "max"]
    ).sort_values("mean", ascending=False)
    print("\nNeighborhood Summary Statistics:")
    print(summary)
    summary["mean"].plot(kind="bar", title="Average Sale Price by Neighborhood", color="skyblue")
    plt.ylabel("Average Sale Price")
    plt.xlabel("Neighborhood")
    plt.tight_layout()
    plt.show()

df.to_csv("RealEstate_Wrangled.csv", index=False)
print("\nWrangled dataset successfully saved to 'RealEstate_Wrangled.csv'")
```
