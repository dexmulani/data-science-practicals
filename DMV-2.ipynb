```python
import requests
from getpass import getpass

API_KEY = getpass("Enter your replacement OpenWeather API key: ").strip()
CITY = "Pune,IN"

response = requests.get(
    "https://api.openweathermap.org/data/2.5/weather",
    params={"q": CITY, "appid": API_KEY, "units": "metric"},
    timeout=20,
)

try:
    data = response.json()
except ValueError:
    raise RuntimeError(
        f"OpenWeather returned HTTP {response.status_code} with a non-JSON response."
    ) from None

if not response.ok:
    message = data.get("message", "Unknown API error")
    # Redact the key if it happens to appear in the response message.
    message = str(message).replace(API_KEY, "[REDACTED]") if API_KEY else str(message)
    raise RuntimeError(
        f"OpenWeather error {response.status_code}: {message}"
    ) from None

record = {
    "City": data["name"],
    "Temperature_C": data["main"]["temp"],
    "Humidity_pct": data["main"]["humidity"],
    "Weather": data["weather"][0]["description"],
    "Wind_Speed_mps": data.get("wind", {}).get("speed"),
}

print(record)
```
